## 2. Load WhatsApp Chat Data

The exported WhatsApp `.txt` file is loaded into Python for further
processing.

```python
file_path = "/content/WhatsApp Chat with Mahaol kendr 🍸.txt"



with open(file_path, "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Total lines:", len(lines))

for line in lines[:10]:
    print(line.strip())
```

```python
messages = []

for line in lines:
  line = line.strip()

  if line == "":
    continue

  print(line)

```

```python
messages = []

for line in lines:
    line = line.strip()

    if line == "":
        continue

    if " - " not in line or ": " not in line:
        continue

    date_time, message_part = line.split(" - ", 1)
    sender, message = message_part.split(": ", 1)

    # Ignore media messages
    if message == "<Media omitted>":
        continue

    # Ignore common WhatsApp system messages
    system_words = [
        "joined using a group link",
        "left",
        "removed",
        "added",
        "created group",
        "changed the subject",
        "changed this group's icon",
        "Messages and calls are end-to-end encrypted",
        "You joined using this group's invite link"
    ]

    is_system_message = False

    for word in system_words:
        if word in message:
            is_system_message = True
            break

    if is_system_message:
        continue

    messages.append({
        "date_time": date_time,
        "sender": sender,
        "message": message
    })

print("Total messages:", len(messages))
print(messages[:5])
```

## 6. NumPy Activity Heatmap

A 6 × 24 NumPy matrix is used to represent group activity across
Monday to Saturday and the 24 hours of the day.

Symbols are used to represent different levels of activity:

- `.` = No activity
- `░` = Low activity
- `▒` = Medium activity
- `█` = High activity

```python
import numpy as np
```

```python
heatmap = np.zeros((6,24),dtype = int)

weekday_index = {
    "Monday": 0,
    "Tuesday": 1,
    "Wednesday": 2,
    "Thursday": 3,
    "Friday": 4,
    "Saturday": 5
}

for msg in messages:
    weekday = msg["weekday"]
    hour = msg["hour"]

    if weekday in weekday_index:
      row = weekday_index[weekday]
      heatmap[row][hour] += 1

print(heatmap)
```

```python
max_value = np.max(heatmap)

print("========== ACTIVITY HEATMAP ==========\n")

print("       ", end="")

for hour in range(24):
    print(f"{hour:02}", end=" ")

print()

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

for i in range(6):
    print(f"{days[i]:<5} ", end="")

    for hour in range(24):
        value = heatmap[i][hour]

        if value == 0:
            symbol = "."
        elif value <= max_value * 0.25:
            symbol = "░"
        elif value <= max_value * 0.60:
            symbol = "▒"
        else:
            symbol = "█"

        print(symbol, end="  ")

    print()
```

```python
stop_words = {
    "the", "is", "a", "an", "and", "to", "of", "in",
    "for", "on", "at", "it", "this", "that", "with",
    "i", "you", "we", "he", "she", "they", "me",
    "my", "your", "are", "was", "were", "be", "am",
    "have", "has", "had", "will", "can", "just"
}

word_count = {}

for msg in messages:
  text = msg["message"].lower()

  for character in ".,!?;:()[]{}\"'":
    text = text.replace(character, " ")

  words = text.split()

  for word in words:
        if word in stop_words:
          continue

        if len(word)<2:
          continue
        if word not in word_count:
          word_count[word] = 0

        word_count[word] +=1
```

## 7. Top 10 Most Frequently Used Words

Messages are converted to lowercase, punctuation is removed, common

```python
person_words = {}

for msg in messages:
    person = msg["sender"]
    text = msg["message"].lower()

    # Remove punctuation
    for character in ".,!?;:()[]{}\"'":
        text = text.replace(character, " ")

    words = text.split()

    if person not in person_words:
        person_words[person] = {}

    for word in words:
        if word in stop_words:
            continue

        if len(word) < 2:
            continue

        if word not in person_words[person]:
            person_words[person][word] = 0

        person_words[person][word] += 1
```

## 9. Response Speed Analysis

Response time is calculated as the time between a message from one
participant and the next message from a different participant.

The average response time is calculated for each participant.

```python
response_times = {}

for msg in messages:
    person = msg["sender"]

    if person not in response_times:
        response_times[person] = []

for i in range(1, len(messages)):
    previous_msg = messages[i - 1]
    current_msg = messages[i]

    previous_person = previous_msg["sender"]
    current_person = current_msg["sender"]

    # Only count when a different person responds
    if previous_person != current_person:
        time_difference = (
            current_msg["date_time"] - previous_msg["date_time"]
        )

        minutes = time_difference.total_seconds() / 60

        response_times[current_person].append(minutes)
```

```python
average_response_time = {}

for person in response_times:
    times = response_times[person]

    if len(times) > 0:
        average_response_time[person] = sum(times) / len(times)
    else:
        average_response_time[person] = 0
```

```python
print("========== AVERAGE RESPONSE TIME ==========")

for person, minutes in average_response_time.items():
    print(person, ":", round(minutes, 2), "minutes")
```

## 10. Inactive Days and Silent Periods

This section identifies the longest consecutive period during which
each participant sent no messages.

The percentage of silent days is also calculated.

```python
active_days = {}

for person in participants:
    active_days[person] = set()

for msg in messages:
    active_days[msg["sender"]].add(msg["date"])
```

```python
from datetime import timedelta

longest_inactive = {}

for person in participants:

    current_streak = 0
    longest_streak = 0

    current_date = start_date

    while current_date <= end_date:

        if current_date not in active_days[person]:
            current_streak += 1

            if current_streak > longest_streak:
                longest_streak = current_streak
        else:
            current_streak = 0

        current_date += timedelta(days=1)

    longest_inactive[person] = longest_streak
```

```python
print("========== INACTIVE STREAKS ==========")

for person, days in longest_inactive.items():
    print(person, ":", days, "days")
```

```python
total_days = (end_date - start_date).days + 1

silent_percentage = {}

for person in participants:
    silent_days = total_days - len(active_days[person])

    silent_percentage[person] = (
        silent_days / total_days
    ) * 100
```

```python
print("========== SILENT DAYS ==========")

for person, percentage in silent_percentage.items():
    print(person, ":", round(percentage, 2), "%")
```

```python
person_stats = {}

for person in participants:
    person_stats[person] = {
        "messages": 0,
        "total_words": 0,
        "night_messages": 0,
        "uppercase_messages": 0,
        "exclamation_messages": 0,
        "question_messages": 0,
        "caring_count": 0,
        "funny_count": 0
    }

print(person_stats)
```

```python
caring_words = {
    "okay",
    "safe",
    "eat",
    "sleep",
    "take care",
    "are you",
    "please",
    "reminder",
    "drink water",
    "don't forget"
}

funny_words = {
    "lol",
    "lmao",
    "haha",
    "rofl",
    "lmfao"
}
```

## 11. Personality Archetype Analysis

The communication patterns of each participant are analyzed using the
archetype rules defined for the GroupDNA project.

```python
for msg in messages:
    person = msg["sender"]
    text = msg["message"]
    lower_text = text.lower()

    person_stats[person]["messages"] += 1

    # Word count
    words = text.split()
    person_stats[person]["total_words"] += len(words)

    # Night messages: 11 PM to 4:59 AM
    hour = msg["hour"]

    if hour >= 23 or hour < 5:
        person_stats[person]["night_messages"] += 1

    # Uppercase messages
    letters = [char for char in text if char.isalpha()]

    if len(letters) > 0:
        uppercase_letters = [
            char for char in letters if char.isupper()
        ]

        uppercase_percentage = (
            len(uppercase_letters) / len(letters)
        )

        if uppercase_percentage > 0.8 and len(letters) >= 3:
            person_stats[person]["uppercase_messages"] += 1

    # Messages with 2 or more !
    if text.count("!") >= 2:
        person_stats[person]["exclamation_messages"] += 1

    # Questions
    if text.rstrip().endswith("?"):
        person_stats[person]["question_messages"] += 1

    # Caring keywords
    for word in caring_words:
        if word in lower_text:
            person_stats[person]["caring_count"] += 1

    # Funny keywords
    for word in funny_words:
        if word in lower_text:
            person_stats[person]["funny_count"] += 1
```

```python
for person in participants:

    total = person_stats[person]["messages"]

    if total > 0:
        person_stats[person]["night_percentage"] = (
            person_stats[person]["night_messages"] / total
        ) * 100

        person_stats[person]["question_percentage"] = (
            person_stats[person]["question_messages"] / total
        ) * 100

        person_stats[person]["uppercase_percentage"] = (
            person_stats[person]["uppercase_messages"] / total
        ) * 100

        person_stats[person]["average_words"] = (
            person_stats[person]["total_words"] / total
        )
    else:
        person_stats[person]["night_percentage"] = 0
        person_stats[person]["question_percentage"] = 0
        person_stats[person]["uppercase_percentage"] = 0
        person_stats[person]["average_words"] = 0
```

```python
for person in participants:
    stats = person_stats[person]

    print("\n", person)
    print("Messages:", stats["messages"])
    print("Average words:", round(stats["average_words"], 2))
    print("Night messages:", stats["night_percentage"], "%")
    print("Uppercase:", stats["uppercase_percentage"], "%")
    print("Questions:", stats["question_percentage"], "%")
    print("Caring count:", stats["caring_count"])
    print("Funny count:", stats["funny_count"])
```

```python
burst_lengths = {}

for person in participants:
    burst_lengths[person] = []

current_person = None
current_burst = 0

for msg in messages:

    person = msg["sender"]

    if person == current_person:
        current_burst += 1

    else:
        if current_person is not None:
            burst_lengths[current_person].append(current_burst)

        current_person = person
        current_burst = 1

# Save the final burst
if current_person is not None:
    burst_lengths[current_person].append(current_burst)
```

```python
average_burst = {}

for person in participants:

    bursts = burst_lengths[person]

    if len(bursts) > 0:
        average_burst[person] = sum(bursts) / len(bursts)
    else:
        average_burst[person] = 0
```

```python
print("========== AVERAGE MESSAGE BURST ==========")

for person, value in average_burst.items():
    print(person, ":", round(value, 2))
```

```python
archetype_scores = {}

for person in participants:
    archetype_scores[person] = {
        "THE SPAMMER": 0,
        "THE GROUP MOM": 0,
        "THE NIGHT OWL": 0,
        "THE STORYTELLER": 0,
        "THE DRAMA QUEEN": 0,
        "THE GHOST": 0,
        "THE COMEDIAN": 0,
        "THE QUESTION MASTER": 0
    }
```

```python
for person in participants:

    stats = person_stats[person]

    # THE SPAMMER
    if average_burst[person] > 3:
        archetype_scores[person]["THE SPAMMER"] = 1

    # THE NIGHT OWL
    if stats["night_percentage"] > 60:
        archetype_scores[person]["THE NIGHT OWL"] = 1

    # THE STORYTELLER
    if stats["average_words"] > 30:
        archetype_scores[person]["THE STORYTELLER"] = 1

    # THE DRAMA QUEEN
    if (
        stats["uppercase_percentage"] > 30
        or stats["exclamation_messages"] >= 2
    ):
        archetype_scores[person]["THE DRAMA QUEEN"] = 1

    # THE GHOST
    if silent_percentage[person] > 60:
        archetype_scores[person]["THE GHOST"] = 1

    # THE QUESTION MASTER
    if stats["question_percentage"] > 25:
        archetype_scores[person]["THE QUESTION MASTER"] = 1
```

```python
highest_caring = max(
    person_stats[person]["caring_count"]
    for person in participants
)
```

```python
for person in participants:

    if person_stats[person]["caring_count"] == highest_caring:
        archetype_scores[person]["THE GROUP MOM"] = 1
```

```python
for person in participants:

    total = person_stats[person]["messages"]

    if total > 0:
        person_stats[person]["funny_percentage"] = (
            person_stats[person]["funny_count"] / total
        ) * 100
    else:
        person_stats[person]["funny_percentage"] = 0
```

```python
highest_funny_percentage = max(
    person_stats[person]["funny_percentage"]
    for person in participants
)
```

```python
for person in participants:

    if (
        person_stats[person]["funny_percentage"]
        == highest_funny_percentage
    ):
        archetype_scores[person]["THE COMEDIAN"] = 1
```

```python
print("========== ARCHETYPE QUALIFICATIONS ==========")

for person in participants:

    print("\n" + person)

    for archetype, score in archetype_scores[person].items():

        if score == 1:
            print("  ✓", archetype)
```

```python
archetype_order = [
    "THE SPAMMER",
    "THE GROUP MOM",
    "THE NIGHT OWL",
    "THE STORYTELLER",
    "THE DRAMA QUEEN",
    "THE GHOST",
    "THE COMEDIAN",
    "THE QUESTION MASTER"
]
```

```python
final_archetypes = {}

for person in participants:

    final_archetypes[person] = "NO ARCHETYPE"

    for archetype in archetype_order:

        if archetype_scores[person][archetype] == 1:
            final_archetypes[person] = archetype
            break
```

```python
print("========== GROUPDNA ARCHETYPES ==========")

for person in participants:

    print(
        person,
        "→",
        final_archetypes[person]
    )
```
